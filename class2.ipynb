```python
# Import libraries
import os
import cv2
import matplotlib
import numpy as np
import matplotlib.pyplot as plt

from zipfile import ZipFile
from urllib.request import urlretrieve

matplotlib.rcParams['figure.figsize'] = (9.0, 9.0)

%matplotlib inline
```

```python
def download_and_unzip(url, save_path):
    print(f"Downloading and extracting assests....", end="")

    # Downloading zip file using urllib package.
    urlretrieve(url, save_path)

    try:
        # Extracting zip file using the zipfile package.
        with ZipFile(save_path) as z:
            # Extract ZIP file contents in the same directory.
            z.extractall(os.path.split(save_path)[0])

        print("Done")

    except Exception as e:
        print("\nInvalid file.", e)
```

```python
URL = r"https://www.dropbox.com/s/48hboi1m4crv1tl/opencv_bootcamp_assets_NB3.zip?dl=1"

asset_zip_path = os.path.join(os.getcwd(), "opencv_bootcamp_assets_NB3.zip")

# Download if assest ZIP does not exists.
if not os.path.exists(asset_zip_path):
    download_and_unzip(URL, asset_zip_path)
```

```python
imagem = cv2.imread("/content/Apollo_11_Launch.jpg")
plt.imshow(imagem)
```

```python


# Define circle parameters
center_x = 200  # x-coordinate of the center
center_y = 300  # y-coordinate of the center
radius = 50    # Radius of the circle
color = (0, 255, 0)  # Green color in BGR (for OpenCV)
thickness = 2   # Thickness of the circle line

# Draw the circle on the image copy to avoid modifying the original
image_with_circle = imagem.copy()
cv2.circle(image_with_circle, (center_x, center_y), radius, color, thickness)

# Display the image with the circle
plt.imshow(image_with_circle)
plt.show()
```

```python
# prompt: Desenhe um retângulo em volta do foguete com open cv


# Define rectangle parameters
x = 500  # x-coordinate of the top-left corner
y = 500  # y-coordinate of the top-left corner
w = 300  # Width of the rectangle
h = 1000  # Height of the rectangle
color = (0, 0, 255)  # Red color in BGR
thickness = 2

# Draw the rectangle on a copy of the image
image_with_rectangle = imagem.copy()
cv2.rectangle(image_with_rectangle, (x, y), (x + w, y + h), color, thickness)

# Display the image with the rectangle
plt.imshow(image_with_rectangle)
plt.show()
```
